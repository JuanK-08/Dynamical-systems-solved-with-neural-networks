```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# ------------ Tiempo de Lyapunov ------------
# Para el clima, se estima aproximadamente en un rango de 10 - 15 días, en segundos:
t_Lyapunov = 15.0 * (3600.0 * 24.0)     # Para 15 días (2 semanas)
```

```python
data_w = pd.read_csv('data_weather.csv', sep=',')

data_w['valid_time'] = pd.to_datetime(data_w['valid_time'])
t_0 = data_w['valid_time'][0]

data_w['time_in_sec'] = (data_w['valid_time'] - t_0).dt.total_seconds()
data_w['time_lyapunov'] = data_w['time_in_sec'] / t_Lyapunov
idx_1liapunov = data_w[data_w['time_lyapunov'] == 1.0].index[0] 

idx_10liapunov = data_w[data_w['time_lyapunov'] == 10.0].index[0]
data_to_import = data_w[0:idx_10liapunov+1]
data_to_import
```

```python
data_to_import.to_excel('data_weather.xlsx', float_format="%.4f")
```

```python
df = pd.read_csv('data_nasa.csv', sep=',', skiprows=14)

df['valid_time'] = pd.to_datetime(
    df[['YEAR', 'MO', 'DY', 'HR']].rename(
        columns={'YEAR': 'year', 'MO': 'month', 'DY': 'day', 'HR': 'hour'})
    )

df = df.drop(columns=['YEAR', 'MO', 'DY', 'HR'])
df.insert(0, 'valid_time', df.pop('valid_time'))


t0 = df['valid_time'][0]

df['time_in_sec'] = (df['valid_time'] - t0).dt.total_seconds()

df['time_lyapunov'] = df['time_in_sec'] / t_Lyapunov

df
```
